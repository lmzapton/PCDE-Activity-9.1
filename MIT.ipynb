# The Massachusetts Institute of Technology (MIT)

**Name:**  Leanne M. Zapton

**GitHub Username:**  lmzapton

The Massachusetts Institute of Technology (MIT) is a private university in Cambridge, Massachusetts. Established in 1861, MIT has since then been at the center of the development of modern technology and science. Since it was founded, MIT has been ranked among the top academic institutions in the world.

The original MIT seal was adopted on December 26, 1864, following a recommendation from a committee that included MIT founder William Barton Rogers.  The seal depicts a scholar holding a book and a worker at an anvil, symbolizing the connection between knowledge and practical application.

**Source:**  Massachusetts Institute of Technology.  History and Symbolism -- MIT Commencement.
